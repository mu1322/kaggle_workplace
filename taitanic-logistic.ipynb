```python
X = train_data[['Pclass', 'Sex', 'Fare']]
y = train_data['Survived']
X
```

```python
#文字列の変換
X = pd.get_dummies(X, columns=['Pclass', 'Sex'])
X
```

```python
#fareのスケーリング
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaler.fit(X)
```

```python
X_scaled = scaler.transform(X)
X = pd.DataFrame(X_scaled, columns=X.columns)
X.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

lr = LogisticRegression()
lr.fit(X_train, y_train)
```

```python
lr.score(X_test, y_test)
```

```python
X_for_submit = test_data[['Pclass', 'Sex', 'Fare']]
X_for_submit = pd.get_dummies(X_for_submit, columns=['Pclass', 'Sex'])
X_for_submit_scaled = scaler.transform(X_for_submit)
X_for_submit_scaled
```

```python
X_for_submit = pd.DataFrame(X_for_submit_scaled, columns=X_for_submit.columns)
X_for_submit
```

```python
X_for_submit['Fare'] = X_for_submit['Fare'].fillna(X_for_submit['Fare'].median())
```

```python
submit = test_data[['PassengerId']].copy()
submit['Survived'] = lr.predict(X_for_submit)
submit
```

```python
submit.to_csv('submission.csv', index=False)
```
